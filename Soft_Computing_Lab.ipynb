# Practical 3: Write a program to implement a Perceptron


```python

import numpy as np

# Training inputs for the AND logic gate
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])

# Expected outputs of the AND gate
y = np.array([0, 0, 0, 1])

# Initialize weights, bias, and learning rate
weights = np.zeros(2)
bias = 0.0
learning_rate = 0.1
epochs = 10

# Train the Perceptron using the training data
for epoch in range(epochs):
    for i in range(len(X)):
        # Calculate weighted sum and add bias
        net_input = np.dot(X[i], weights) + bias

        # Apply the step activation function
        prediction = 1 if net_input > 0 else 0

        # Calculate prediction error
        error = y[i] - prediction

        # Update weights and bias using the Perceptron learning rule
        weights += learning_rate * error * X[i]
        bias += learning_rate * error

# Test the trained Perceptron
print("Trained Weights:", weights)
print("Trained Bias:", round(bias, 2))
print("\nAND Gate Predictions:")

for sample in X:
    net_input = np.dot(sample, weights) + bias
    prediction = 1 if net_input > 0 else 0
    print(sample, "->", prediction)
```

# Practical 5: Write a script to implement the ADALINE model

```python
import numpy as np

# Define input values and their corresponding target outputs
X = np.array([0, 1, 2, 3], dtype=float)
y = np.array([1, 3, 5, 7], dtype=float)

# Initialize weight, bias, and learning rate
weight = 0.0
bias = 0.0
learning_rate = 0.01
epochs = 1000

# Train the ADALINE model using the LMS learning rule
for epoch in range(epochs):
    for i in range(len(X)):
        # Calculate the linear output
        output = weight * X[i] + bias

        # Calculate the error between target and predicted output
        error = y[i] - output

        # Update weight and bias using the LMS rule
        weight += learning_rate * error * X[i]
        bias += learning_rate * error

# Predict outputs using the trained model
predictions = weight * X + bias

# Display the trained parameters and predictions
print("Trained Weight:", round(weight, 4))
print("Trained Bias:", round(bias, 4))
print("\nTarget and Predicted Outputs:")

for i in range(len(X)):
    print("Input:", X[i],
          "| Target:", y[i],
          "| Predicted:", round(predictions[i], 2))
```

# Practical 6: Write a Python module to implement a Multilayer Neural Network

```python
import numpy as np

# XOR input and output data
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([[0], [1], [1], [0]])

# Initialize weights and biases
np.random.seed(42)
W1 = np.random.randn(2, 4)
W2 = np.random.randn(4, 1)
b1 = np.zeros((1, 4))
b2 = np.zeros((1, 1))
lr = 1.0

# Sigmoid activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Train the neural network
for _ in range(10000):
    h = sigmoid(X @ W1 + b1)
    out = sigmoid(h @ W2 + b2)

    # Calculate errors and update parameters
    error = out - y
    dW2 = h.T @ error / len(X)
    db2 = np.mean(error, axis=0, keepdims=True)

    hidden_error = (error @ W2.T) * h * (1 - h)
    dW1 = X.T @ hidden_error / len(X)
    db1 = np.mean(hidden_error, axis=0, keepdims=True)

    W2 -= lr * dW2
    b2 -= lr * db2
    W1 -= lr * dW1
    b1 -= lr * db1

# Display final predictions
h = sigmoid(X @ W1 + b1)
out = sigmoid(h @ W2 + b2)

print("XOR Gate Predictions:")
for inputs, result in zip(X, (out >= 0.5).astype(int)):
    print(inputs, "->", result[0])
```

# Practical 8: Write a script to implement the Union of Fuzzy Sets

```python
# Define two fuzzy sets
A = {1: 0.2, 2: 0.5, 3: 0.7}
B = {1: 0.6, 2: 0.3, 3: 0.9}

# Calculate the union using maximum membership values
union = {x: max(A[x], B[x]) for x in A}

# Display the fuzzy sets
print("Fuzzy Set A:", A)
print("Fuzzy Set B:", B)
print("Union of A and B:", union)
```

# Practical 9: Write a program to implement a Fuzzy Model

```python
# Define the temperature
temperature = 28

# Calculate fuzzy membership values
warm = max(0, min(1, (temperature - 20) / 10))
hot = max(0, min(1, (temperature - 25) / 10))

# Define fan speeds for the fuzzy rules
warm_speed = 50
hot_speed = 100

# Calculate the final fan speed using weighted average
fan_speed = (warm * warm_speed + hot * hot_speed) / (warm + hot)

# Display the results
print("Temperature:", temperature, "°C")
print("Warm Membership:", round(warm, 2))
print("Hot Membership:", round(hot, 2))
print("Fan Speed:", round(fan_speed, 2), "%")
```

# Practical 12: Write a Python script to implement Principal Component Analysis (PCA) for dimensionality reduction

```python
from sklearn.decomposition import PCA
import numpy as np

# Create a dataset with two features
X = np.array([[2, 4], [3, 5], [4, 7], [5, 8]])

# Apply PCA to reduce two features to one
pca = PCA(n_components=1)
reduced_data = pca.fit_transform(X)

# Display the results
print("Original Data Shape:", X.shape)
print("Reduced Data:\n", np.round(reduced_data, 2))
print("Reduced Data Shape:", reduced_data.shape)
```

# Practical 14: Write a Python script to implement K-means clustering

```python
from sklearn.cluster import KMeans

# Define the dataset
X = [[1, 2], [2, 3], [3, 2],
     [8, 8], [9, 8], [8, 9]]

# Create and train the K-means model
model = KMeans(n_clusters=2, random_state=42, n_init=10)
model.fit(X)

# Display cluster labels and centroids
print("Cluster Labels:", model.labels_)
print("Cluster Centers:\n", model.cluster_centers_)
```

# Practical 15: Write a Python script to implement Hierarchical clustering

```python
from sklearn.cluster import AgglomerativeClustering

# Define the dataset
X = [[1, 2], [2, 3], [3, 2],
     [8, 8], [9, 8], [8, 9]]

# Create and apply the clustering model
model = AgglomerativeClustering(n_clusters=2)
labels = model.fit_predict(X)

# Display the results
print("Cluster Labels:", labels)
```
