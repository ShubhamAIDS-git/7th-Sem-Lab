**Question:** Calculate the mean, median, and mode for a given dataset.

```python
# Import NumPy library for numerical calculations
import numpy as np

# Import the stats module from SciPy for statistical functions
from scipy import stats

# Create the dataset
data = [10, 15, 20, 20, 25, 30, 35, 20, 40, 40]

# Calculate the mean (average) of the dataset
mean_value = np.mean(data)

# Calculate the median (middle value) of the dataset
median_value = np.median(data)

# Calculate the mode (most frequently occurring value)
# keepdims=True keeps the result in array form
mode_value = stats.mode(data, keepdims=True)

# Display the dataset
print("Dataset:", data)

# Display the calculated mean
print("Mean:", mean_value)

# Display the calculated median
print("Median:", median_value)

# Display the calculated mode
# [0] extracts the first mode value from the result
print("Mode:", mode_value.mode[0])
```

# Practical 5

**Question:** Use the binomial distribution to model a probability scenario.

```python
# Import NumPy for numerical calculations
import numpy as np

# Define the number of trials
# Here, a coin is tossed 10 times
n = 10

# Define the probability of success in one trial
# For a fair coin, probability of getting heads is 0.5
p = 0.5

# Define the number of successes we want
# Here, we want exactly 6 heads
k = 6

# Calculate the probability of getting exactly k successes
# np.math is not used because NumPy does not provide it in recent versions
# The binomial probability formula is:
# P(X = k) = C(n,k) * p^k * (1-p)^(n-k)

from math import comb

probability = comb(n, k) * (p ** k) * ((1 - p) ** (n - k))

# Display the result
print("Number of trials:", n)
print("Probability of success:", p)
print("Required number of successes:", k)
print("Probability of exactly 6 heads:", probability)
print("Probability in percentage:", probability * 100, "%")
```

## Viva Questions & Answers

**Q1. What is a binomial distribution?**  
Binomial distribution is a probability distribution used to model the number of successes in a fixed number of independent trials.

**Q2. What are the conditions for a binomial distribution?**

**Q3. What are the two possible outcomes in a binomial experiment?**

**Q4. What is the real-world scenario used in this practical?**  
A coin is tossed 10 times and we calculate the probability of getting exactly 6 heads.

**Q5. What is the value of n in this practical?**  
`n = 10`, representing 10 coin tosses.

**Q6. What is the value of p?**  
`p = 0.5`, because the probability of getting heads with a fair coin is 0.5.

**Q7. What is k?**  
`k` represents the required number of successes. Here, `k = 6`.

**Q8. What is the probability of exactly 6 heads?**  
The probability is **0.205078125**, or approximately **20.51%**.

**Q9. What is the formula of binomial probability?**  
`P(X = k) = C(n,k) × p^k × (1-p)^(n-k)`

**Q10. What does `comb(n, k)` calculate?**  
It calculates the number of combinations of selecting `k` successes from `n` trials.

**Q11. What does `p ** k` represent?**  
It represents the probability of getting success `k` times.

**Q12. What does `(1-p) ** (n-k)` represent?**

**Q13. Why is the probability of success 0.5?**  
Because a fair coin has two equally likely outcomes: heads and tails.

**Q14. Where is binomial distribution used in data science?**  
It can be used in quality control, A/B testing, survey analysis, medical studies, and classification of binary outcomes.

```python
# Import the norm function from SciPy
# It is used to work with the normal probability distribution
from scipy.stats import norm

# Define the mean of the dataset
mean = 70

# Define the standard deviation
standard_deviation = 10

# Define the observed value
# We want to find the probability of getting a score below 80
x = 80

# Calculate the z-score
# Formula: Z = (X - Mean) / Standard Deviation
z_score = (x - mean) / standard_deviation

# Calculate the cumulative probability
# norm.cdf() gives the probability of getting a value
# less than or equal to the given value
probability = norm.cdf(z_score)

# Display the results
print("Mean:", mean)
print("Standard Deviation:", standard_deviation)
print("Observed Value:", x)
print("Z-Score:", z_score)
print("Probability of score below 80:", probability)
print("Probability in percentage:", probability * 100, "%")
```

# Practical 8

**Question:** Calculate the margin of error and construct a confidence interval.

```python
# Import NumPy for numerical calculations
import numpy as np

# Import stats from SciPy for statistical functions
from scipy import stats

# Sample dataset
data = [68, 72, 75, 70, 74, 78, 71, 73, 76, 69]

# Get the number of observations in the sample
n = len(data)

# Calculate the sample mean
mean = np.mean(data)

# Calculate the sample standard deviation
# ddof=1 is used for sample standard deviation
std = np.std(data, ddof=1)

# Calculate the critical t-value for a 95% confidence level
t_value = stats.t.ppf(0.975, n - 1)

# Calculate the margin of error
margin_error = t_value * (std / np.sqrt(n))

# Calculate the lower and upper limits of the confidence interval
lower = mean - margin_error
upper = mean + margin_error

# Display the results
print("Sample Mean:", mean)
print("Margin of Error:", margin_error)
print("95% Confidence Interval:", (lower, upper))
```

**Q1. What is a confidence interval?**

**Q3. What is the formula for a confidence interval?**

# Practical 9

**Question:** Perform a hypothesis test and interpret the results.

```python
# Import NumPy for numerical calculations
import numpy as np

# Import the t-test function from SciPy
from scipy import stats

# Sample dataset
data = [68, 72, 75, 70, 74, 78, 71, 73, 76, 69]

# Set the hypothesized population mean
# H0: Population mean = 70
hypothesized_mean = 70

# Perform a one-sample t-test
t_statistic, p_value = stats.ttest_1samp(data, hypothesized_mean)

# Set the significance level
alpha = 0.05

# Display the test results
print("Sample Mean:", np.mean(data))
print("t-statistic:", t_statistic)
print("p-value:", p_value)

# Interpret the hypothesis test
if p_value < alpha:
    print("Result: Reject the null hypothesis.")
else:
    print("Result: Fail to reject the null hypothesis.")
```

## Viva Questions & Answers

**Q1. What is hypothesis testing?**  
Hypothesis testing is a statistical method used to determine whether there is sufficient evidence to support a claim about a population.

**Q2. What is the null hypothesis?**  
The null hypothesis (H₀) is the statement that there is no significant difference or effect.

**Q3. What is the alternative hypothesis?**  
The alternative hypothesis (H₁) is the statement that there is a significant difference or effect.

**Q4. What is the null hypothesis in this practical?**  
H₀: The population mean is equal to 70.

**Q5. What is the significance level used here?**  
The significance level is **0.05 or 5%**.

**Q6. What is a p-value?**  
A p-value indicates how much evidence there is against the null hypothesis.

**Q7. What is the decision rule based on p-value?**  
If p-value < 0.05, reject H₀. Otherwise, fail to reject H₀.

**Q8. What is the p-value in this practical?**  
It is approximately **0.028**.

**Q9. What is the conclusion of this test?**  
Since p-value < 0.05, we reject the null hypothesis.

**Q10. Which statistical test is used?**  
A one-sample t-test is used.

**Q11. What does `stats.ttest_1samp()` do?**  
It performs a one-sample t-test to compare the sample mean with a hypothesized population mean.

**Q12. What is the t-statistic?**  
The t-statistic is a value used to measure the difference between the sample mean and the hypothesized mean relative to the variability in the sample.

**Q13. Which libraries are used in this practical?**  
NumPy and SciPy are used.

**Q14. What does `alpha = 0.05` represent?**  
It represents the chosen significance level of 5%.

**Q15. What does "fail to reject H₀" mean?**  
It means there is not enough statistical evidence to reject the null hypothesis.

# Practical 11

**Question:** Apply ARIMA modeling to make predictions in a time series dataset.

```python
# Import NumPy for numerical calculations
import numpy as np

# Import ARIMA model from Statsmodels
from statsmodels.tsa.arima.model import ARIMA

# Create a simple time series dataset
data = [10, 12, 15, 14, 18, 20, 19, 23, 25, 24,
        28, 30, 29, 32, 35, 34, 38, 40, 39, 42]

# Create and fit the ARIMA model
# order=(p,d,q): p = autoregression, d = differencing, q = moving average
model = ARIMA(data, order=(1, 1, 1))
model_fit = model.fit()

# Forecast the next 3 values
forecast = model_fit.forecast(steps=3)

# Display the forecasted values
print("Forecasted values:", forecast)
```

## Viva Questions & Answers

**Q1. What is ARIMA?**  
ARIMA stands for AutoRegressive Integrated Moving Average. It is a statistical model used for time series forecasting.

**Q2. What is time series forecasting?**

**Q3. What are the three components of ARIMA?**  
AR (AutoRegressive), I (Integrated), and MA (Moving Average).

**Q4. What does AR represent?**  
AR represents the relationship between the current value and previous values of the time series.

**Q5. What does I represent?**  
I represents differencing used to make a non-stationary time series more stationary.

**Q6. What does MA represent?**  
MA represents the relationship between the current value and previous forecast errors.

**Q7. What does `order=(1,1,1)` mean?**  
It means p=1, d=1, and q=1.

**Q8. What does p represent?**  
p represents the number of autoregressive terms.

**Q9. What does d represent?**  
d represents the number of times differencing is applied.

**Question:** Evaluate the accuracy of time series forecasts using appropriate metrics.

```python
# Import NumPy for numerical calculations
import numpy as np

# Actual and forecasted values
actual = np.array([100, 120, 130, 150, 170])
forecast = np.array([105, 115, 128, 145, 175])

# Calculate Mean Absolute Error (MAE)
mae = np.mean(np.abs(actual - forecast))

# Calculate Mean Squared Error (MSE)
mse = np.mean((actual - forecast) ** 2)

# Calculate Root Mean Squared Error (RMSE)
rmse = np.sqrt(mse)

# Calculate Mean Absolute Percentage Error (MAPE)
mape = np.mean(np.abs((actual - forecast) / actual)) * 100

# Display the results
print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("MAPE:", mape, "%")
```

**Question:** Apply Bayes' Theorem to update probabilities based on new information.

```python
# Import Bayes' theorem related calculation tools
from scipy import stats

# Probability of having a disease
P_D = 0.01

# Probability of testing positive when a person has the disease
P_positive_given_D = 0.95

# Probability of testing positive when a person does not have the disease
P_positive_given_not_D = 0.05

# Probability of not having the disease
P_not_D = 1 - P_D

# Calculate the overall probability of a positive test
P_positive = (
    P_positive_given_D * P_D
    + P_positive_given_not_D * P_not_D
)

# Apply Bayes' Theorem
P_D_given_positive = (
    P_positive_given_D * P_D
) / P_positive

# Display the result
print("Prior Probability of Disease:", P_D)
print("Probability of Positive Test:", P_positive)
print("Posterior Probability of Disease:", P_D_given_positive)
print("Posterior Probability in percentage:",
      P_D_given_positive * 100, "%")
```
